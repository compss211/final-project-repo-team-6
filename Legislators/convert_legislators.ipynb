```python
import json

with open('../legislators-historical.json', 'r') as f:
    data = json.load(f)

# Check first legislator
first_legislator = data[0]
print("First legislator structure:")
print(f"Name: {first_legislator.get('name', {})}")
print(f"Number of terms: {len(first_legislator.get('terms', []))}")
print(f"First term: {first_legislator.get('terms', [{}])[0]}")
```
